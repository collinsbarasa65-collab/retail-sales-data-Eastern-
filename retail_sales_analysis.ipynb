# Stage 6 RFM Customer Segmentation

```python
# Task 1

# Compute Recency: days since each customer's last purchase.
# import thew dependencies
import pandas as pd
import numpy as np
from datetime import datetime

# Convert 'InvoiceDate' to datetime format
retail_sales_data['InvoiceDate'] = pd.to_datetime(retail_sales_data['InvoiceDate'])

# Calculate the last purchase date for each customer
last_purchase_date = retail_sales_data.groupby('CustomerID')['InvoiceDate'].max().reset_index()

# Define a reference date (e.g., the day after the latest transaction in the dataset)
# or the current date if you want real-time recency.
# For consistency with the dataset, let's use the day after the last transaction.
# This assumes 'InvoiceDate' contains datetime objects after previous conversion.
current_date = retail_sales_data['InvoiceDate'].max() + pd.Timedelta(days=1)

# Calculate Recency in days
last_purchase_date['Recency'] = (current_date - last_purchase_date['InvoiceDate']).dt.days

# Display the recency results
print(last_purchase_date[['CustomerID', 'Recency']].head())
```

```python
# Task 2 & 3 combined
snapshot = retail_sales_data['InvoiceDate'].max() + pd.Timedelta(days=1)
rfm = retail_sales_data.groupby('CustomerID').agg(
    Recency = ('InvoiceDate', lambda x:
    (snapshot - x.max()).days))
```

```python
rfm = retail_sales_data.groupby('CustomerID').agg(
    Recency = ('InvoiceDate', lambda x: (snapshot - x.max()).days),
    Frequency = ('InvoiceNo', 'nunique'), # Count unique invoice numbers for frequency
    Monetary = ('Revenue', 'sum'))# Sum of revenue for monetary value
rfm.head() # selected the first 5 rows
```

```python
# Task 4

# Score each metric from 1–4 using pd.qcut().

# First ,to compute the  quartile value for each column and name then R , F , M.

# Recency quartile

r_labels = range(4, 0,-1)
r_quartiles = pd.qcut(rfm['Recency'], 4, labels = r_labels)
retail_sales_data = rfm.assign(R = r_quartiles.values)
retail_sales_data.head()

```

```python
# TAsk 4...cont

# Frequency and monetary quartiles
f_labels = range(1,5)
m_labels = range(1,5)
f_quartiles = pd.qcut(rfm['Frequency'], 4, labels = f_labels)
m_quartiles = pd.qcut(rfm['Monetary'], 4, labels = m_labels)
retail_sales_data = rfm.assign(F = f_quartiles.values)
retail_sales_data = rfm.assign(M = m_quartiles.values)
retail_sales_data.head()
```

```python
#Task 5...cont
# Build RFM score
# sum all RFM quartiles values to RFM_Score
# assign R,F,M scores to the dataframe (rfm)

rfm['R'] = r_quartiles.values
rfm['F'] = f_quartiles.values
rfm['M'] = m_quartiles.values

def join_rfm(x):
  return str(x['R']) + str(x['F']) + str(x['M'])

rfm['RFM_Score'] = rfm[['R','F','M']].sum(axis=1)
rfm.head()
```

```python
# Task 6

# build a rfm segment
seg_map = {
    r'[1-2][1-2]': 'hibernating',
    r'[1-2][3-4]': 'at_Risk',
    r'[1-2]5': 'cant_loose',
    r'3[1-2]': 'about_to_sleep',
    r'33': 'need_attention',
    r'[3-4][4-5]': 'loyal_customers',
    r'41': 'promising',
    r'51': 'new_customers',
    r'[4-5][2-3]': 'potential_loyalists',
    r'5[4-5]': 'champions'
}

# Create an RF_String by concatenating R and F scores to match seg_map patterns
rfm['RF_String'] = rfm['R'].astype(str) + rfm['F'].astype(str)
rfm['segment'] = rfm['RF_String'].replace(seg_map, regex=True)

def join_rfm(x): return str(x['R']) + str(x['F']) + str(x['M'])
rfm['RFM Segment'] = rfm.apply(join_rfm, axis=1)
rfm
```

```python
# Data visualisation Barplot

import matplotlib.pyplot as plt
import seaborn as sns

# Count of customers in each segment
segment_counts = rfm['segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Number of Customers']

# plot the points
plt.figure(figsize=(10, 6))
sns.barplot(x='Number of Customers', y='Segment', data=segment_counts, palette='magma', hue='Segment', legend=False)
plt.title('Distribution of Customers Across RFM Segments',fontsize=10,color='r')
plt.xlabel('Number of Customers')
plt.ylabel('Customer Segment')
plt.tight_layout()
plt.savefig('Chart6.png',dpi=150)
plt.show()

print('Customer Segment Distribution:')
print(segment_counts)
```

```python
# Task 2

# A review of Kenya's mean revenue in contrast to Uganda,Tanzania and Rwanda

# using label based indexing(loc/df.loc) I select each country and its mean

kenya_mean_revenue = country_stats.loc['Kenya', 'Mean']
rwanda_mean_revenue = country_stats.loc['Rwanda', 'Mean']
tanzania_mean_revenue = country_stats.loc['Tanzania', 'Mean']
uganda_mean_revenue = country_stats.loc['Uganda', 'Mean']

print(f"Mean Revenue for Kenya: {kenya_mean_revenue:.2f}")
print(f"Mean Revenue for Rwanda: {rwanda_mean_revenue:.2f}")
print(f"Mean Revenue for Tanzania: {tanzania_mean_revenue:.2f}")
print(f"Mean Revenue for Uganda: {uganda_mean_revenue:.2f}")

print(f"\nKenya's mean revenue is {kenya_mean_revenue - rwanda_mean_revenue:.2f} higher than Rwanda.")
print(f"Kenya's mean revenue is {kenya_mean_revenue - tanzania_mean_revenue:.2f} higher than Tanzania.")
print(f"Kenya's mean revenue is {kenya_mean_revenue - uganda_mean_revenue:.2f} higher than Uganda.")
```

Task 3

```python
# Use scipy.stats.ttest_ind() with alternative='greater'. This is to performs a one-tailed independent two-sample t-test to determine if the mean of the first sample is statistically larger than the mean of the second sample

# Import the dependency
import scipy.stats as stats
import pandas as pd

# Load retail_sales_data and re-calculate 'Revenue' to ensure 'Country' column is present
retail_sales_data_original = pd.read_csv('/content/retail_sales_data.csv')
retail_sales_data_original['Revenue'] = retail_sales_data_original['Quantity'] * retail_sales_data_original['UnitPrice']

# assigning group_a sample to Kenya revenue
# assigning  group_b sample to other countries' revenue)according to raw data
group_a = retail_sales_data_original[retail_sales_data_original['Country'] == 'Kenya']['Revenue']
group_b = retail_sales_data_original[retail_sales_data_original['Country'] != 'Kenya']['Revenue']

# one-tailed t-test implementation
res = stats.ttest_ind(group_a, group_b, alternative='greater')
print(f"T-statistic: {res.statistic:.2f}")
print(f"P-value:     {res.pvalue:.2f}")
```

A P-Value ≥ 0.05 suggests that the evidence is not strong enough to reject the null hypothesis. Our P-value is greater than 0.05 hence there's insufficient significant proof to conclusively state that the mean revenue for Kenya is statistically greater than the mean revenue of all other countries combined. While the mean revenue for Kenya was numerically higher, this difference is not considered statistically significant at the 0.05 level.

On the other hand, the T-statistic of 1.26, in conjunction with the P-value of 0.10, does not provide strong evidence against the null hypothesis. The observed difference in means, while positive, is not statistically significant enough to confidently conclude that Kenya's mean revenue is greater than that of other countries based on this data. Both the T-statistic and P-value consistently indicate insufficient proof to assert that Kenya's mean revenue is significantly greater than her peers.

Limitations of the Retail Sales Dataset

* Limited Timeframe Review Period: The dataset covers a specific period (e.g., one year).This one year period  might not conclusively be a  representation of long-term trends as it fails to review seasonal variations outside this period, or customer behavior changes over a longer duration of time ,say, 5 years

* Macroeconomic Factors: The dataset doesn't account for external factors that could influence sales such as the individual countries inflation and interest rates,countries GDP ,disposable income level, unforseen market turbulences (e.g. pandemics, supply chain disruptions) that all play a role in determining the purchasing power. Devoid of such context, interpretation of sales fluctuations poses challenges.
